#### Задача 4 (*) - Манхэттенское расстояние.

Мы уже много чего делали с косинусным сходством. Давайте теперь рассмотрим еще один подход к вычислению семантической близости между словами - *геометрические метрики*. В отличие от косинусного сходства, которое измеряет только угол (направление) векторов, эти метрики смотрят на физическое расстояние между точками. Существуют две основные метрики:

**Евклидово расстояние** (L2-норма)

• Что это: Обычное расстояние между двумя точками «по прямой», как если бы мы приложили линейку на карте.

• В лингвистике: Используется редко в чистом виде, так как сильно зависит от частоты слов. Популярные слова имеют более длинные векторы и улетают на периферию пространства. Из-за этого евклидово расстояние между «кошка» и «кот» может быть больше, чем между редкими словами из разных тем. Но если векторы заранее нормализовать (привести к одной длине), евклидово расстояние становится математически эквивалентным косинусному.

**Манхэттенское расстояние** (L1-норма / Расстояние городских кварталов)

• Что это: Расстояние, которое нужно пройти не по прямой, а строго по сетке осей (как такси по улицам Нью-Йорка).

• В лингвистике: Очень полезно в пространствах высокой размерности (например, 4096 признаков в LLM). Оказывается, при росте количества признаков евклидово расстояние начинает «размываться» (все точки кажутся одинаково удаленными), а Манхэттенское расстояние гораздо лучше сохраняет контраст между похожими и непохожими словами.

Давайте посмотрим, как работает Манхэттенское расстояние, на примере. Представьте, что мы создаем умную систему, например, для зоопарка.

####Задача 4 (*) - Дамерау-Левенштейн.

Расстояние **Дамерау-Левенштейна** — это расширение классического расстояния Левенштейна, которое добавляет четвёртую операцию редактирования: транспозицию (перестановку) двух соседних символов. Оно точнее отражает реальные опечатки при наборе текста на клавиатуре, где люди часто меняют соседние буквы местами.

Напишите функцию поиска расстояния Дамерау-Левенштейна в парадигме динамического программирования.

**Важно:**

Алгоритм Дамерау-Левенштейна существует в двух вариантах.

1. Ограниченное расстояние (Optimal String Alignment, OSA). Запрещает изменять уже переставленную подстроку повторно. Например, переход CA $\rightarrow $ ABC займёт 3 операции вместо 2 (CA $\rightarrow $ AC $\rightarrow $ ABC). Реализуется простым добавлением одного условия в матрицу ДП (проверка s1[i-1] == s2[j-2] и s1[i-2] == s2[j-1]).
2. Истинное расстояние Дамерау-Левенштейна. Разрешает редактировать символы после перестановки. Для его вычисления уже недостаточно смотреть только на клетки (i-1, j-1) и (i-2, j-2) — требуется хранить словарь последних вхождений каждого символа алфавита.

Для реализации я вам рекомендую первый вариант.

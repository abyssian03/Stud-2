## Регулярные выражения

Регулярное выражение - это шаблон, который соотносится с последовательностью символов в тексте. По существу, РВ - это обобщенная запись для разных вариантов строк, например, можно обобщить, как выглядят все телефонные номера в тексте или все e-mail адреса. РВ придумали в 1950-х гг., когда математик Стефен Коул Клини формализовал описание регулярного языка. Звездочка Клини называется в его честь.

### Что нужно знать про РВ



**Когда их использовать**

Если вы уверены, что без РВ не обойтись. Прежде чем писать РВ, подумайте, нельзя ли решить задачу стандартными методами строк или множеств? РВ - мощный инструмент, но для некоторых задач это overkill. У РВ есть своя вычислительная сложность: для компиляции регулярного выражения длины n компьютеру приходится тратить $2^n$ времени. А еще в РВ легко запутаться, даже опытному программисту бывает не так-то просто составить правильный шаблон, который не захватывал бы лишнего и не пропускал бы нужного.

Но вообще РВ очень часто используются для:
- поиска по корпусам (в ГИКРЯ поддерживаются некоторые возможности РВ)
- поиска по текстам внутри файлов (на серверах - но даже ваша привычная Windows кое-что знает про РВ, хотя у нее свои правила)
- токенизации, сегментации по предложениям
- обработки путей файлов
...

###Regex в Python

За работу с регулярками в Python отвечает модуль re. Его не нужно устанавливать, так как он входит в стандартную библиотеку.

```python
import re
```

Одно из первых применений РВ - это поиск по текстам, содержащимся внутри файлов. До 90-х годов прошлого века операционные системы не имели графического интерфейса, а во многих файлах лежала текстовая информация. В 70-з гг. программисты создали программу grep, которая с помощью РВ ищет все файлы, внутри которых есть совпадающие тексты. grep работает из командной строки в системах unix (linux, MacOS) и до сих пор активно используется при работе с серверами.

У РВ есть несколько типов синтаксиса (flavors), которые различаются функциональностью. Самые известные такие:

- POSIX (имеет два варианта: базовый и расширенный, BRE & ERE)
- Perl 5 (разработчики языка Perl добавили свои плюшки)
- Perl 6 = Raku Rules
- PCRE (Perl Compatible Regular Expressions)

Некоторые ЯП имеют свои версии синтаксиса РВ, в основном очень похожие на PCRE; питон в их числе.

Центральные объекты модуля re - это классы re.Pattern и re.Match.

Класс Pattern - это внутреннее питонье представление вашего шаблона. Регулярное выражение в питоне - это всегда строка, которую модуль re превращает в свой собственный объект.

Класс Match - это класс того, что нашлось на ваше регулярное выражение. Он содержит несколько вещей:

1. Само найденное выражение + отдельно содержимое групп РВ, если они там были.
2. Индексы найденного выражения в исходной строке.

####Методы

У объектов класса Match есть три метода поиска совпадений:

```python
m.start()  # возвращает индекс исходной строки, с которого вхождение начинается
m.end()  # то же, только для конца
n.span()  # возвращает кортеж из этих индексов
```



Чтобы в классе Match найти все выражение целиком, нужно использовать метод

Если нужно достать только содержимое какой-то из групп, то можно передать ее номер:

```python
m.group(1)
```

С ролью и использованием групп мы подробно познакомимся ниже.

####Компиляция

Любая функция модуля re, прежде чем искать по регулярному выражению, должна его *скомпилировать*. Функции re умеют компилировать регулярки сами, но можно скомпилировать строку с рв заранее:

```python
pattern = re.compile(r'\d+')
```

Зачем это нужно: предобработка шаблона в байт-код для ускорения при многократном использовании.


####Основные функции модуля re

  - re.match(pattern, string, flags) - ищет совпадение только в начале строки, возвращает объект Match или None.
  - re.search(pattern, string, flags) - ищет совпадение по всей строке, возвращает первое.
  - re.fullmatch(pattern, string, flags) - проверяет, что строка полностью подходит под паттерн.
  - re.findall(pattern, string, flags) - возвращает все совпадения (строки). Важный нюанс: если в шаблоне есть группы, вернет список кортежей групп.
  - re.finditer(pattern, text, flags) - то же, что findall, только возвращает итератор из объектов класса match. Можно превратить в список, а можно сразу итерироваться.
  - re.sub(pattern, sub, string, flags) - заменяет все вхождения pattern на строку sub (вместо строки можно передать функцию, которая принимает объект класса match и возвращает строку).
  - re.split(pattern, string) - разбивает строку по шаблону.


  

```python
print(re.search(r"py", "python .py"))
for p in re.finditer(r"py", "python .py"):
  print(p)
print(re.sub(r"py", "cy", "python .py"))
```

flags в каждой из этих функций - необязательный параметр. Вы можете передать флаги, если хотите, например,  игнорировать регистр - за это отвечает флаг

re.IGNORECASE (= re.I)

О других флагах можете посмотреть документацию.

Флаги можно также указывать внутри регулярного выражения с помощью (?): r'(?i)...' - так передается флаг ignorecase.

### Правила составления шаблонов

###Основной синтаксис

Регулярное выражение читается слева направо и ищет посимвольно. Буквы и цифры обычно соответствуют сами себе.

То есть, допустим, мы хотим найти слово из трех букв. Мы можем представить его как

    _ _ _
    
Дальше уже нужно подумать, что поставить на место каждого прочерка. Если мы хотим искать конкретные буквы, то можем прямо их и написать, например,

    _ _ _
    c a t
    
Найдет нам все вхождения "cat" в тексте (ну или хотя бы одно, в зависимости от функции).

```python
print(re.match(r"cat", "catapult"))
```

####Метасимволы и специальные последовательности

Эти символы нельзя использовать как обычный текст без подготовки:

\ . ^ $ * + ? { } [ ] ( ) |

Это такие символы, которые означают не сами себя, а что-то еще.
- \ - работает как символ экранирования (как и в Python)
- . - обозначает любой символ вообще (кроме \\n)
- ^ и $ означают начало и конец строки.
- \+, \*, ?, {} - квантификаторы (см. ниже)
- | - обозначает оператор "или" (логическое объединение)
- [] - обозначают класс
- () - обозначают группу


Бэкслеш отключает специальное значение символа; raw-строка r"..." сохраняет его для RegEx.

То есть, если вам нужно искать сами эти символы, то их нужно экранировать с помощью бэкслеша: \\. позволит искать точку, например.

```python
print(re.search(r"\.", "python .py"))
```

Сам бэкслеш тоже, конечно, нужно экранировать: \\\\

**Специальные последовательности** мы уже знаем из обычного питона: \\n и \\t. То есть, специальная последовательность - это какая-то буковка (или набор их) после бэкслеша. РВ распознают и стандартные питоновские последовательности, но имеют и некоторые свои. Какие нам пригодятся:

- \\d - любая цифра (те символы, которые в юникоде имеют ярлык "digit")
- \\w - любая цифра или буква (те символы, которые в юникоде считаются цифробуквенными). Примечание: туда еще входит нижнее подчеркивание _
- \\s - любой пробельный символ (пробел, неразрывный пробел, табуляция, перенос на новую строку...)

\* Любую из этих последовательностей можно использовать наоборот, если писать заглавную букву. То есть, \\D означает все символы, кроме цифр.


```python
re.search(r"admin_...", "admin_123 is my admin")
```

Еще последовательности:

- \A и \Z означают абсолютное начало и конец всего текста, соответственно.
- \\b - граница слова: это не символ, а условие, которое говорит, что на этом месте должно быть \\W\\w или наоборот.

\* \\B означает, наоборот, позицию внутри слова.

Обратите внимание: в Python строка \b — это спецсимвол Backspace, а в RegEx — граница слова. Поэтому всегда пишите префикс r перед кавычками шаблона!



```python
print(re.split(r"\b", "*python*is*the*best*"))
```

####Квантификаторы

Мы можем искать телефонные номера с помощью \\d, например. Предположим, наши номера состоят из шести любых цифр, разделенных дефисами:

    _ _ - _ _ - _ _
    
Мы можем записать такие номера РВ как:

    \d \d - \d \d - \d \d
    
Но если мы хотим одиннадцатизначные номера, в которых цифры просто идут подряд? Или все номера, где от 6 до 11 цифр?

Такие вещи позволяют искать квантификаторы: они сообщают поисковому движку, что искать нужно не один символ, а сколько-то таких символов подряд. Тогда:

    _ _ _ _ _ _ _ _ _ _ _ - наш одиннадцатизначный номер превратится в
    \d{11}
    
Итак, квантификаторы применяются к одному символу, стоящему слева от них. Какие они бывают:

- {n} велит искать ровно n таких символов
- {n,m} велит искать не меньше n и не больше m символов
- {n,} велит искать не меньше n, но чем больше, тем лучше
- {,m} - велит искать от 0 до m
- ? - либо один такой символ, либо ни одного сойдет
- \* - либо 0 таких символов, либо чем больше, чем лучше
- \+ - от 1 до бесконечности символов!

Таким образом:

\\d{3} - найдет 3 цифры

\\d{2,3} - найдет либо 2, либо 3 цифры, но охотнее 3

\\d{2,} - найдет минимум 2 цифры, но лучше побольше

\\d{,3} - найдет от 0 до 3 цифр

\\d? - либо найдет пустоту, либо 1 цифру

\\d\* - либо найдет пустоту, либо как можно больше цифр

\\d+ - найдет 1 и больше цифр

```python
re.findall(r"1|0+", "10110010001")
```

Такие квантификаторы называются жадными: они ищут как можно больше повторяющихся символов. Можно сделать квантификатор ленивым, чтобы он искал поменьше, если поставить сразу после него ?, например, \\d+?, если сможет, будет стараться обходиться только одной цифрой.

```python
re.findall(r"1|0+?", "10110010001")
```

####Классы символов

Квадратные скобки обозначают классы символов. Класс символов - это некоторый набор символов, которые нас устраивают в этом месте. Мы уже знаем некоторые классы: \\d, \\w, \\s. Но можно задавать и свои, произвольные. Например, мы хотим найти в тексте все слова these, но those нас в принципе тоже устраивают. Тогда:

    _ _ _ _ _
    t h [eo] s e
    
То есть, внутри квадратных скобок можно просто перечислить все символы, которые нам нравятся. Сами квадратные скобки будут занимать **только одну позицию**: они обозначают один символ, любой из перечисленных внутри них.

Такие классы удобны, например, для подсчета альтернативных словоформ.

```python
text="Бабушка достала из буфета пачку белого сахара и аккуратно насыпала в чашку пару ложек ароматного сахару, чтобы чай стал по-настоящему сладким"
print(sum(1 for _ in re.finditer(r"сахар[уа]", text)))
```

Внутри квадратных скобок в основном метасимволы теряют свою силу, и . делается обычной точкой, например. Но некоторые ее, наоборот, приобретают или изменяют.

Дефис, если не стоит с краю, обозначает диапазон: [0-9] - это все цифры от 0 до 9. Как берется диапазон? По таблице юникода. Мы уже знаем, что в основном буквы там идут по алфавиту, но буква ё, например, убежала не туда. Значит, чтобы захватить весь русский алфавит с ё, нужно писать: [а-яё]. А если нас еще и заглавные буквы интересуют, то придется написать [А-ЯЁа-яё].

^ из начала строки превращается в отрицание, но только в начале класса. Если нас интересует **все**, кроме того, что мы перечислили, можно записать: [^а-яё]. Такая запись будет означать "все, кроме кириллических строчных букв".

Не теряет свою силу только \\, потому что внутри класса мы можем написать \\d или \\t, например.

```python
print(re.sub(r"[0-9]", "", "A1B2C3"))
```

####Группы
Группы нужны в основном для каких-то технических целей. Самая, пожалуй, частая - это для того, чтобы ограничить действие оператора "или".

По умолчанию "или" работает на все РВ целиком: он считает, что нас устраивает или все, что стоит слева, или все, что стоит справа. То есть,

    a|bc будет искать либо a, либо bc.
    
Но если мы хотим искать либо ac, либо bc? Ну, можно написать явно: ac|bc, но часто повторяющаяся часть оказывается слишком длинной. Тут на помощь нам приходят скобки, которые выделят символы в группу и ограничат |:

    (a|b)c = ac|bc
    
Группы также можно использовать для того, чтобы ссылаться на них. Группы в РВ нумеруются по левой скобке.

**Пример:** Мы ищем дату 2026-11-10. Шаблон r"\d{4}-\d{2}-\d{2}" найдет строку целиком.

Если мы расставим скобки: r"(\d{4})-(\d{2})-(\d{2})":

• Группа 0 (всегда): вся строка 2026-11-10

• Группа 1: год 2026

• Группа 2: месяц 11

• Группа 3: день 10

    

```python
match = re.search(r"(\d{4})-(\d{2})-(\d{2})", "Дата: 2026-11-10")
if match:
    print(match.group(0))  # "2026-11-10" (все совпадение)
    print(match.group(1))  # "2026" (только первая группа)
    print(match.groups())  # ('2026', '11', '10') (все группы кортежем)
```

Также есть возможность создавать **именованные группы**.

Именованные группы делают регулярные выражения гораздо более читаемыми, избавляя от необходимости считать скобки по порядку (group(1), group(2)...).
В Python для создания именованной группы используется синтаксис (?P<имя_группы>...).

```python
text = "Пользователь admin зашел с IP-адреса 192.168.1.50"
pattern = r"Пользователь (?P<user>\w+) зашел с IP-адреса (?P<ip>[\d.]+)"
match = re.search(pattern, text)
if match:
    username = match.group("user")
    ip_address = match.group("ip")
    print(f"Имя пользователя: {username}")
    print(f"IP-адрес: {ip_address}")
```

Если у нас есть именованные группы, мы можем легко превратить их в словарь с помощью метода .groupdict():

```python
data_dict = match.groupdict()
print("Готовый словарь:", data_dict)
```

Иногда, правда, группа нам нужна не для того, чтобы на нее ссылаться, и тогда используются группы без захвата содержимого. Это не сбивает нумерацию групп.

    (?:...) # группирует без сохранения

```python
match = re.findall(r"(?:ha)+", "ha-a-haha!")
print(match)
```

Конструкция \number: Позволяет потребовать, чтобы в этом месте строки стоял точно такой же текст, какой был пойман ранее группой.

Например, вот так можно найти повторяющиеся слова, разделенные пробелом:

```python
text = "hello hello world"
match = re.search(r"(\w+) \1", text)
print(match.group(1))
```

**Проверки**

Проверки - это уже достаточно хитрое средство РВ (в POSIX их нет, их придумали создатели Perl). Проверка - это когда мы хотим проверить какое-то условие, чтобы слева или справа от нашего искомого текста были определенные вещи, при этом включать их в результат поиска не хотим. Проверки бывают 4 видов:

- опережающая позитивная (positive lookahead) = (?=...) - смотрим справа от того, что ищем, и берем только в том случае, если условие проверки выполняется
- опережающая негативная (negative lookahead) = (?!...) смотрим справа от того, что ищем, и берем только в том случае, если условие проверки НЕ выполняется
- ретроспективная позитивная (positive lookbehind) = (?<=...) - смотрим слева от того, что ищем, и берем только в том случае, если условие проверки выполняется
- ретроспективная негативная (negative lookbehind) = (?<!...) - смотрим слева от того, что ищем, и берем только в том случае, если условие проверки НЕ выполняется

Как это понимать? Например, мы хотим найти в тексте имя "Вася", но только такое, чтобы после него шла фамилия "Пупкин". Тогда нам нужна опережающая позитивная проверка:

    Вася (?=Пупкин)
    
А если нас интересуют все Васи, кроме Пупкина, это будет:

    Вася (?!=Пупкин)
    
Если же мы китайцы и пишем сперва фамилию, а потом имя, и при этом нас тоже интересует Вася, то:

    (?<=Пупкин) Вася - найдет всех Пупкиных Василиев
    (?<!Пупкин) Вася - найдет всех Василиев, кроме Пупкиных
    
Внутри проверки может быть тоже какой-то регулярный шаблон, и проверки даже бывают вложенные, но лучше с ними не перебарщивать, а то легко запутаться. :)


Вот пример поиска цифр с помощью ретроспективной проверки на символ #:

```python
text = "ID: #99 и №10"
print(re.findall(r"(?<=#)\d+", text))
```

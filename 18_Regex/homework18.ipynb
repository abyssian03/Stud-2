#### Задача 1

Напишите функцию для поиска дат в разных форматах (ДД.ММ.ГГГГ, ГГГГ-ММ-ДД) и нормализации их к единому стандарту с помощью re.sub.

#### Задача 2

Российский автомобильный номер (частных лиц) имеет строго определенный формат: 1 буква, 3 цифры, 2 буквы. Но есть нюанс: в номерах используются только те кириллические буквы, которые имеют графические аналоги в латинице (чтобы иностранцы могли их прочесть). Это 12 букв: А, В, Е, К, М, Н, О, Р, С, Т, У, Х.

Вам нужно создать валидатор номеров, который извлечет из текста все правильные номера и проигнорирует ошибочные.

**Пример:**

*Вход:*

"На парковку заехали машины: А123ВЕ77, Я999ОХ99, К555НТ199, Щ111ЩЩ77."

*Выход:*

['А123ВЕ77', 'К555НТ199']

#### Задача 3

Представьте, что у вас есть длинный текст, в котором множество раз встречается конструкция "ни Х, ни У" (она также может быть "ни Х, ни У, ни Z" и даже "ни Х, ни У, ни Z, ни Q").

Вам нужно составить функцию, которая для каждого вхождения такой конструкции посчитает, сколько в ней элементов перечисления. То есть для "ни Петю, ни Васю" ваша функция должна вернуть 2, для "ни на английском, ни на французском, ни на немецком" — 3.

#### Задача 4

Напишите самый простой токенизатор: он должен захватывать стандартные знаки препинания, русские и английские слова (символы с диакритиками не берем в расчет). Сам токенизатор должен быть функцией, которая возвращает список словарей, у которых есть ключи "форма" (собственно текст) и "тип": пунктуация или слово. За слово (англ или рус) считаем любую последовательность букв одного алфавита, внутри которой может быть один или несколько дефисов, при этом дефисы не могут быть с краев и идти подряд.

**Усложнение:** Подумайте, как может быть реализована очистка от мусорных символов, если они стоят внутри слова.

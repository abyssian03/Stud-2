## Графы: Структуры и обходы

### Определение графа, его виды и сферы применения

####Определение

**Граф** — это абстрактная модель, которая описывает систему объектов и связи между ними.

Формально граф состоит из двух основных множеств:

• Множество **вершин** (узлов, точек), которые представляют сами объекты.

• Множество **рёбер** (связей, линий), соединяющих эти вершины.



**Теория графов** — раздел дискретной математики, изучающий графы, одна из ветвей топологии.

Теория графов имеет корни в работе Леонарда Эйлера 1736 года, посвященной решению задачи о кёнигсбергских мостах.

В 1936 году вышла первая в мире книга по теории графов (на немецком языке) Денеша Кёнига «Теория конечных и бесконечных графов» с большей частью результатов за прошедшие 200 лет. В частности, в книге Кёнига имеется общий параграф для задачи о кёнигсбергских мостах и задаче о домино (построение замкнутой цепи из всех костей домино по правилам игры).

Сегодня передовые исследования в рамках теории графов в математике связаны с [Теоремой о четырех красках](https://ru.wikipedia.org/wiki/%D0%A2%D0%B5%D0%BE%D1%80%D0%B5%D0%BC%D0%B0_%D0%BE_%D1%87%D0%B5%D1%82%D1%8B%D1%80%D1%91%D1%85_%D0%BA%D1%80%D0%B0%D1%81%D0%BA%D0%B0%D1%85). Это была первая крупная математическая задача, решенная с помощью компьютерного доказательства (доказана в 1976 году Кеннетом Аппелем и Вольфгангом Хакеном из Иллинойского университета).

####Основные понятия

Для нас будут важны следующие понятия:

• **Путь** (цепь) — последовательность рёбер, ведущая от одной вершины к другой.

• **Цикл** — замкнутый путь, начинающийся и заканчивающийся в одной и той же вершине.

• **Длина пути** (или цикла) — число составляющих его рёбер.

• **Петля** — цикл единичной длины, т.е. ребро, которое соединяет вершину саму с собой.

• **Кратные ребра** — несколько рёбер, соединяющих одну и ту же пару вершин.

Если граф не имеет петель и кратных ребер, он называется **простым**. Граф с петлями, но без кратных ребер, называется **псевдографом**, граф с кратными ребрами, но без петель - **мультиграфом**, а с тем и другим - **псевдомультиграфом**.

Если граф не имеет циклов, он называется **ациклическим**, иначе - **циклическим**.

По характеру ребер различают **ненаправленные графы** (те, в которых ребра представляют собой симметричные связи), **направленные** или **ориентированные графы** (те, в которых ребра представляют связи по принципу перехода или подчинения, изображаются на диаграммах стрелками) и **смешанные графы** (те, в которых присутствуют ребра обоих типов).

Также выделяют понятие **взвешенного графа**, в котором каждому ребру приписывается определенное числовое значение - вес.

По характеру связности различают **связные графы** (те, в которых для любых вершин u, v есть путь из u в v), **не связные графы** и **сильно** или **ориентированно связные** (связные графы, где из любой вершины в любую другую есть **ориентированный путь**).

#### Графы в лингвистике

В лингвистике графы позволяют формализовать структуру языка, представляя слова, звуки или смыслы как вершины, а отношения между ними — как рёбра. Это основа для компьютерной лингвистики, машинного перевода и поисковых систем.

Как применяются в различных областях лингвистики?

1. **Синтаксис.**

- Деревья зависимостей: вершины — это слова в предложении, а направленные рёбра показывают, какое слово от какого зависит (например, от глагола идёт стрелка к подлежащему и дополнению). Такое представление помогает алгоритмам понимать структуру фразы, даже если в языке свободный порядок слов.

- Деревья составляющих: Группируют слова в более крупные блоки (именная группа, глагольная группа), показывая иерархию предложения.

2. **Семантика и лексикология.**

- Тезаурусы и формальные онтологии: Вершинами выступают понятия (синсеты), а рёбрами — смысловые отношения: синонимия, антонимия, гипонимия («является разновидностью», например: пудель $\rightarrow $ собака $\rightarrow $ животное).

- Графы знаний: Используются поисковиками и голосовыми помощниками для понимания фактов (например, Пушкин — [автор] $\rightarrow$ Евгений Онегин).

3. **Морфология и фонетика.**

- Конечные автоматы для распознавания слов: Путь по ориентированному графу от начальной вершины к конечной позволяет отделить приставку, корень, суффикс и окончание, чтобы привести слово к начальной форме.

- Конечные трансдьюсеры для распознавания звучащей речи: Акустические модели используют графы, где вершины — это фонемы, а веса рёбер — вероятность перехода одного звука в другой.

4. **Дистрибутивная семантика и анализ текстов.**

- Графы совместной встречаемости слов: Слова соединяются ребром, если они часто стоят рядом в одном предложении или абзаце. Анализ таких графов (поиск кластеров и центральных вершин) помогает автоматически выделять главные темы текста и ключевые слова.

- Дискурсивные деревья, графы сущностей, графы аргументации, графы диалогов: Используются для распознавания логической структуры текста.

### Создание графа вручную в Python



#### Связь с ДП

На прошлой лекции мы решали задачи динамического программирования (например, считали расстояние Левенштейна).

На самом деле, таблица ДП — это **ориентированный ациклический граф (DAG)**, где:
- Вершины — это состояния (клетки таблицы).
- Рёбра — это переходы между состояниями.

**В чём проблема ДП для языка?**
ДП работает только там, где нет циклов (мы всегда движемся вперёд: от меньших подстрок к большим). Но естественный язык полон циклов:
- синонимы ссылаются друг на друга (*дом - здание - постройка - дом*);
- в тексте мы возвращаемся к уже упомянутым словам и темам;
- в толковых словарях встречаются порочные круги в определениях.

Чтобы алгоритмы не уходили в бесконечный цикл, нам нужна теория графов.

#### Как хранить простой не взвешенный граф?

Предположим, у нас есть словарь из n = 50 000 слов.
1. **Матрица смежности** (двумерный массив):
- Займёт $2.5 \times 10^9$ ячеек памяти (примерно 20 ГБ RAM для float64), хотя большинство слов может быть не связано друг с другом.
- Поиск всех соседей слова занимает O(n).
- Если есть циклы (а в простом графе могут быть циклы длиной больше 1), структура сразу ломается.

2. **Список смежности** (словарь):
- Храним только существующие связи. Память: O(n + e), где e — количество рeбер.
- Проверка «связаны ли слово А и слово Б» работает за O(1) при использовании поиска по ключу.
- Структура допускает циклы.

**Вывод:** Список смежности предпочтителен, по крайней мере, для "разреженного графа" (более строго: графа, матрица смежности которого разрежена) и графа с циклами.

####DefaultDict для списков смежности

Очень удобно составлять список смежности с использованием специализированной структуры данных модуля collections под названием **defaultdict**, которая представляет собой просто словарь с дефолтным значением для любого нового ключа.

Главное преимущество defaultdict перед обычным dict — отсутствие ручных проверок на наличие ключа и защита от ошибки KeyError при построении и обходе графа.

Рассмотрим пример. Пусть нам дан список пар (гипоним, гипероним):

```python
relations = [
    ("такса", "собака"),
    ("пудель", "собака"),
    ("собака", "млекопитающее"),
    ("кот", "млекопитающее"),
    ("млекопитающее", "животное"),
    ("воробей", "птица"),
    ("птица", "животное"),
]
```

Соберем ориентированный граф для этого списка с помощью обычного dict.

```python
graph = {}
for u, v in relations:
    if u not in graph:
        graph[u] = []
    graph[u].append(v)

print(graph)
```

А теперь проделаем то же самое с помощью defauldict.

```python
import collections
```

```python
defaulted_graph = collections.defaultdict(list)
for u, v in relations:
    defaulted_graph[u].append(v)

print(defaulted_graph)
```

Попробуем обратиться к отсутствующему ключу defaultdict и увидим, что программа не упадет:

```python
print(defaulted_graph['животное'])
```

Ненаправленный не взвешенный граф можно хранить точно так же, но надо следить за тем, чтобы при добавлении каждого нового слова-ключа это слово добавлялось в список значений к каждому слову-ключу из его собственного списка.

#### Как хранить взвешенный граф?

В лингвистике связи часто имеют вес: частотность биграммы, вероятность перехода или семантическую близость.
Такой граф удобно хранить как словарь словарей.

Ниже приведен пример создания взвешенного ориентированного графа колокаций (совместной встречаемости слов) и поиска данных в нем. Вершины — слова, а вес ребра — сколько раз словосочетание встретилось в корпусе (частотность).

```python
collocations = {
    "высокий": {
        "человек": 15,
        "дом": 42,
        "уровень": 120,
        "риск": 85
    },
    "уровень": {
        "жизни": 95,
        "воды": 30,
        "шума": 25
    },
    "риск": {
        "ошибки": 40,
        "потери": 18
    }
}

# Узнаем вес конкретного ребра:
print("Частота 'высокий уровень':", collocations["высокий"]["уровень"])

# Посмотрим все продолжения слова "уровень" и их веса:
for next_word, weight in collocations["уровень"].items():
    print(f"уровень -> {next_word} (вес: {weight})")
```

Обратите внимание, что поиск по-прежнему работает за O(1).

Имея такой граф, можно посмотреть, например, относительную частоту (вероятность) появления словосочетания "высокий уровень жизни".

```python
def phrase_probability(graph, phrase):
    words = phrase.split()
    prob = 1.0  # Нейтральный элемент для умножения

    # Идём по парам соседних слов (по рёбрам пути)
    for w1, w2 in zip(words, words[1:]): # заканчиваем работу, когда кончается один из списков

        if w1 not in graph or w2 not in graph[w1]:
            # Раз слова нет в графе, вероятность приравниваем к нулю
            return 0.0

        edge_weight = graph[w1][w2] # вес конкретного ребра
        total_out_weight = sum(graph[w1].values())  # сумма весов всех выходящих рёбер из w1

        step_prob = edge_weight / total_out_weight # считаем условную вероятность для пары слов
        print(f"P('{w2}' | '{w1}') = {edge_weight} / {total_out_weight} = {step_prob:.3f}")

        prob *= step_prob

    return prob

p = phrase_probability(collocations, "высокий уровень жизни")
print(f"\nИтоговая относительная частота (вероятность) фразы: {round(p, 3)} или {round(p * 100, 1)}%")
```

### Базовый обход графа: Поиск в ширину (BFS)

Главная задача на графах: **найти путь от вершины А до вершины Б** (и желательно кратчайший).

Для этого используется **BFS (Breadth-First Search)**:
1. Используем очередь **collections.deque** (так как popleft() работает за O(1), в отличие от list.pop(0) за O(n)).
2. Используем множество visited = set(), чтобы не проверять одну вершину дважды и не попасть в цикл.

#### Пошаговый алгоритм BSF:
1. **Подготовка (инициализация):**
   * Создаём очередь — список слов, которые нам ещё предстоит проверить. Сразу кладём туда стартовое слово `start`.
   * Создаём множество посещённых вершин `visited` — нашу «записную книжку», куда мы записываем слова сразу, как только поставили их в очередь (чтобы не добавить одно слово дважды). Кладём туда `start`.

2. **Главный цикл (пока очередь не пуста):**
   * Достаём самое первое слово из начала очереди (queue.popleft()). Назовём его `current`.
   * **Проверка цели:** Если `current` совпадает с `target`, мы нашли путь! Останавливаемся и возвращаем True (или количество шагов).

3. **Осмотр соседей:**
   * Берём всех прямых соседей слова `current` из графа.
   * Для каждого соседа проверяем: есть ли он уже в `visited`?
     * Если да — пропускаем его (мы там уже были или он уже стоит в очереди).
     * Если нет — записываем его в `visited` и добавляем в **конец** очереди.

4. **Финал:**
   * Если очередь опустела, а слово `target` так и не встретилось — значит, пути между словами не существует. Возвращаем False.

Попробуем запрограммировать этот алгоритм на примере графа синонимов. Сначала введем данные для обработки.

```python
# Создаем ненаправленный граф. Обратите внимание на цикл: дом - здание - постройка - дом
synonyms = {
    "дом": ["здание", "жильё", "постройка"],
    "здание": ["дом", "постройка", "сооружение"],
    "постройка": ["дом", "здание"],
    "жильё": ["дом", "обитель", "квартира"],
    "обитель": ["жильё", "приют", "хижина"],
    "квартира": ["жильё"],  # тупиковая ветка
    "хижина": ["обитель", "изба"],
    "изба": ["хижина"],
    "сооружение": ["здание"],
    "глагол": ["слово"]      # изолированная компонента (пути от "дом" сюда нет)
    "слово": ["глагол"]
}
```

```python
А теперь напишем саму функцию обхода графа.
```

```python
def can_reach_bfs(graph, start, target):
    # Инициализация очереди и множества посещённых вершин
    queue = collections.deque([start])
    visited = {start}

    print(f"Старт поиска от '{start}' к '{target}':\n")

    # Главный цикл
    while queue:
        current = queue.popleft()
        print(f"Проверяем слово: '{current}' | Очередь сейчас: {list(queue)}")

        # Условие успешного завершения поиска
        if current == target:
            return True

        # Просматриваем всех соседей текущего слова
        for neighbor in graph.get(current, []): # По умолчанию список пуст
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                print(f"Добавили новое слово '{neighbor}'")

    # Условие неуспешного завершения
    return False

if can_reach_bfs(synonyms, "дом", "хижина"):
    print(f"\nУспех! Слово найдено.")
else:
    print(f"\nПуть не найден.")
```

Тот же самый алгоритм, но только с сохранением в очереди всего пути, а не отдельного слова, можно использовать для выдачи самой итоговой цепочки до слова `target`. Вот так:

```python
def find_path_bfs(graph, start, target):
    queue = collections.deque([[start]])
    visited = {start}

    print(f"Старт поиска от '{start}' к '{target}':\n")

    while queue:
        path = queue.popleft()
        # Здесь мы достаём целый путь, а текущее слово - это его последний элемент
        current = path[-1]

        if current == target:
            return path  # Возвращаем найденный маршрут

        for neighbor in graph.get(current, []):
            if neighbor not in visited:
                visited.add(neighbor)
                # Создаём новый список-путь, добавляя соседа в конец
                new_path = path + [neighbor]
                # Метод .append() изменяет исходный список на месте, а нам может быть нужно два независимых ответвления маршрута.
                # Операция + создаёт новый список, не портя старый, поэтому делаем через нее.

                queue.append(new_path)

    return None  # Если пути нет, возвращаем None (вместо False)

print("Путь:", find_path_bfs(synonyms, "дом", "хижина"))
```

### Второй базовый алгоритм: Поиск в глубину (DFS)

Если **BFS** (поиск в ширину) расходится кругами — сначала проверяет всех ближайших соседей, потом соседей соседей, и т.д. — то **DFS (Depth-First Search)** работает как исследователь в лабиринте: он идёт по одной ветке до самого тупика, а затем возвращается к последней развилке.

#### Где это полезно в лингвистике?
- Обход деревьев зависимостей (синтаксический разбор предложения) или деревьев составляющих.
- Поиск всех потомков (гипонимов) в семантических сетях (например, найти все подклассы класса *животное* в биологической таксономии).
- Проверка графа на наличие циклов (например, порочных кругов в словарных определениях).

#### Пошаговый алгоритм (рекурсивный вариант):
В основе DFS лежит **стек** (принцип LIFO — «последним пришёл, первым ушёл»). Самый простой способ написать DFS в Python — использовать **рекурсию**, где роль стека выполняет стек вызовов самой программы.

1. **Вход в вершину:**
   * Мы заходим в текущую вершину `current`.
   * Сразу записываем её в множество посещённых `visited`, чтобы не зациклиться.
   * Выполняем полезное действие (например, печатаем слово или добавляем его в список результатов).

2. **Шаг вглубь (рекурсия):**
   * Берём список соседей `current`.
   * Для каждого соседа проверяем: есть ли он уже в `visited`? Если нет — сразу вызываем для него эту же функцию. Программа «проваливается» глубже, откладывая остальных соседей текущей вершины на потом.

3. **Возврат (откат / backtracking):**
   * Когда у вершины не остаётся непосещённых соседей (мы дошли до «листа» дерева или тупика), функция завершается, и мы автоматически возвращаемся на шаг назад — к предыдущей развилке.

Попробуем это закодить. Так же сначала вводим данные.

```python
# Создаем ориентированный граф.
taxonomy = {
    "существо": ["животное", "растение"],
    "животное": ["млекопитающее", "птица"],
    "млекопитающее": ["собака", "кот"],
    "собака": ["такса", "пудель"],
    "такса": [],
    "пудель": [],
    "кот": [],
    "птица": ["воробей"],
    "воробей": [],
    "растение": ["дерево", "цветок"],
    "дерево": ["дуб"],
    "дуб": [],
    "цветок": []
}
```

Теперь пишем саму функцию обхода.

```python
def dfs_hyponyms(graph, current, visited=None, depth=0):
    # Инициализируем множество visited при самом первом запуске
    if visited is None:
        visited = set()
        print(f"Иерархия подклассов класса '{current}':\n")

    # Помечаем вершину как посещённую и выводим её с отступом (для наглядности глубины)
    visited.add(current)
    print("  " * depth, current)

    # Идём вглубь по каждому непосещённому соседу
    for neighbor in graph.get(current, []):
        if neighbor not in visited:
            #Рекурсивный вызов (погружаемся на уровень глубже)
            dfs_hyponyms(graph, neighbor, visited, depth + 1)

dfs_hyponyms(taxonomy, "животное")
```

#### Итеративный DFS: магия одной строчки

Рекурсивный DFS красив, но в Python у него есть ограничение на глубину рекурсии. Поэтому на практике DFS часто пишут **без рекурсии** — точно так же, как BFS, но вместо **очереди** (deque) берут обычный **список** (list):

- В **BFS** мы берём элемент **сначала** очереди: .popleft() (принцип FIFO: первым пришёл — первым ушёл).
- В **DFS** мы берём элемент **с конца** списка: .pop() (принцип LIFO: последним пришёл — первым ушёл).

Закодим этот вариант на нашем примере с таксономией. Для разнообразия будем выводить не всю иерархию подклассов, а путь от верхнего таксона к нижнему.

```python
def find_path_dfs(graph, start, target):
    # В стеке, как и в очереди для BFS, храним не отдельные слова, а целые пути
    stack = [[start]]
    visited = {start}

    while stack:
        # Достаём путь с КОНЦА стека (принцип LIFO — идём вглубь)
        path = stack.pop()
        current = path[-1]

        if current == target:
            return path

        for neighbor in graph.get(current, []):
            if neighbor not in visited:
                visited.add(neighbor)
                new_path = path + [neighbor]
                stack.append(new_path)

    return None

path = find_path_dfs(taxonomy, "существо", "кот")
if path:
    print("Путь найден:", " -> ".join(path))
else:
    print("Пути нет.")
```

###Что делать, если в графе есть цикл?

Представим фрагмент толкового словаря с «порочным кругом»:
*сепулька -> объект -> сепуление -> процесс -> сепулька*

Что произойдет с нашими алгоритмами обхода вглубь, когда они наткнутся на такой круг?

А ничего страшного! Вот что конкретно:

1. Когда мы доходим до слова *процесс*, его сосед — *сепулька*.
2. Мы проверяем: `if neighbor not in visited:`.
3. Так как *сепулька* уже лежит в *visited*, мы **не добавляем её снова**, просто игнорируем это ребро и безопасно завершаем обход.

*Но что если наша задача именно в том, чтобы найти такой цикл, когда он есть?*

В этом случае нам нужен более сложный алгоритм, помечающий вершины графа в зависимости от того, в каком состоянии они находятся.

Он называется **алгоритмом трех цветов**.

#### Как просто понять «алгоритм трёх цветов»? (Метафора открытых вкладок)

Представьте, что вы проверяете толковый словарь, переходя по ссылкам в определениях слов (как в Википедии).

Чтобы не запутаться, у вас есть два списка (множества `set`):
1. **`checking_now` («Открытые вкладки» / Серый цвет):** слова, которые вы открыли прямо сейчас и ещё не дочитали до конца их цепочку определений.
2. **`checked_done` («Прочитано и закрыто» / Чёрный цвет):** слова, которые вы уже полностью проверили — там точно нет порочных кругов.

**Как поймать цикл?**
* Если вы переходите по ссылке на слово из **`checked_done`** — всё отлично, туда можно даже не заходить, там всё чисто.
* Если вы переходите по ссылке на слово, которое лежит в **`checking_now`** — **СТОП! Вы наткнулись на свою же открытую вкладку!** Цепочка замкнулась сама на себя.

Попробуем закодит этот алгоритм для графа определений, содержащего цикл:


```python
dictionary_graph = {
    "сепулька": ["объект", "вещь"],
    "вещь": ["предмет"],
    "предмет": [],
    "объект": ["сепуление"],
    "сепуление": ["процесс"],
    "процесс": ["сепулька", "действие"],  # <-- Ребро обратно в "сепульку" замыкает цикл!
    "действие": []
}
```

```python
# Рекурсивная функция - проверяет цепочку определений, начиная с ОДНОГО конкретного слова
def has_cycle_from_word(graph, word, checking_now, checked_done):
    # Открываем вкладку с текущим словом (красим в "серый")
    checking_now.add(word)

    # Смотрим на все слова в его определении
    for next_word in graph.get(word, []):

        # Если ссылка ведёт в уже открытую вкладку — мы поймали цикл!
        if next_word in checking_now:
            print(f"Пойман порочный круг: '{word}' ссылается на уже открытое слово '{next_word}'!")
            return True

        # Если слово ещё не проверено полностью, вызываем функцию для него
        if next_word not in checked_done:
            if has_cycle_from_word(graph, next_word, checking_now, checked_done):
                return True

    # Если все ссылки из этого слова проверены, и циклов нет,
    # закрываем вкладку (убираем из "серых") и отправляем в архив ("чёрные")
    checking_now.remove(word)
    checked_done.add(word)
    return False


# Главная функция — запускает проверку для всего словаря целиком
def check_full_dictionary(graph):
    checking_now = set()  # Открытые вкладки (серые)
    checked_done = set()  # Полностью проверенные слова (чёрные)

    for word in graph:
        # Запускаем проверку только для тех слов, которые ещё не в архиве
        if word not in checked_done:
            if has_cycle_from_word(graph, word, checking_now, checked_done):
                return True  # Как только нашли хотя бы один цикл — останавливаемся

    return False


check_full_dictionary(dictionary_graph)
```

### Бонус: Что делать, если у шагов разная цена? (Алгоритм Дейкстры)

Сегодня мы научились обходить граф в ширину с помощью BFS и в глубину с помощью DFS.

BFS помогает нам найти кратчайший путь до нужного нам слова, DFS позволяет увидеть структуру графа и выловить циклы.

Но у базового алгоритма BFS есть важное ограничение: он считает, что любой переход стоит одинаково (1 шаг).

А что, если у рёбер есть **вес** (например, семантическое расстояние между словами или «стоимость» редкого перехода в корпусе)?

Тогда нам поможет следующий алгоритм:

#### Алгоритм Дейкстры (Вспоминаем жадные алгоритмы!)
По сути, Дейкстра - это **BFS + жадный выбор**:
1. Вместо обычной очереди (`deque`), где слова идут строго по порядку добавления, используется **очередь с приоритетом** — из неё всегда первым достаётся слово с **минимальной накопленной суммой весов** от старта.
2. Если до какого-то слова нашёлся более дешёвый обходной маршрут, алгоритм обновляет его лучшую стоимость («релаксация ребра»).

#### Зачем это нужно в лингвистике?
* Поиск самой естественной смысловой цепочки в графах эмбеддингов.
* Поиск самой вероятной последовательности слов или частей речи.

**Пример:**
Представим, что вес ребра — это **«редкость»** (семантическая удалённость) перехода между словами: чем меньше число, тем естественнее связь.
* Прямой переход: *кот* $\xrightarrow{\text{вес } 10}$ *мышь* (1 шаг, но редкая прямая связка)
* Обходной путь: *кот* $\xrightarrow{\text{вес } 2}$ *ловит* $\xrightarrow{\text{вес } 3}$ *мышь* (2 шага, но оба перехода очень естественные)

**Как сработают алгоритмы?**
- BFS выберет прямой путь *кот -> мышь*, потому что он просто не смотрит на веса.
- Алгоритм Дейкстры выберет путь *кот -> ловит -> мышь*, потому что его суммарная стоимость меньше.

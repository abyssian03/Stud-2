Разберите данный ниже код для решения следующей задачи:

Имеется поле n × m (n - количество строк, m - количество столбцов), сколькими способами можно добраться королём из нижней левой клетки до правой верхней клетки.

В комментариях напишите, как он работает и насколько, на ваш взгляд, он эффективен (с обоснованием). Подумайте, как его можно оптимизировать.

```python
def king_moves(n,m):
    king_set = [[0] * m for i in range(n)]
    for i in range(n):
        for j in range(m):
            if i == 0 and j == 0:
                king_set[0][0] = 1
            elif i == 0:
                king_set[0][j] = 1
            elif j == 0:
                king_set[i][0] = 1
            else:
                king_set[i][j] = king_set[i-1][j] + king_set[i][j-1]
            print(king_set)
    return king_set[n-1][m-1]

def king_moves_2(n,m):
    king_set = []
    for i in range(n):
        for j in range(m):
            if i == 0 and j == 0:
                king_set.append(1)
            elif i == 0:
                king_set.append(1)
            elif j == 0:
                king_set.append(1)
            else:
                king_set.append(king_set[-1] + king_set[-n])
            print(king_set)
    return king_set[n*m-1]

print(king_moves(3,3))
print(king_moves_2(3,3))
```

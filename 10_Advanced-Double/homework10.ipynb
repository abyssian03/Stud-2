**Рекурсия** - метод программирования, при котором функция внутри своего тела вызывает сама себя для вычисления своего параметра. У любой правильной рекурсивной функции есть две обязательные части:

1. База рекурсии (условие выхода) - простая ситуация, при которой функция не вызывает себя, а сразу возвращает готовый результат. Без него программа войдет в бесконечное повторение и зависнет или выдаст ошибку переполнения памяти.

2. Шаг рекурсии - вызов функцией самой себя с новыми, более простыми или меньшими аргументами, которые приближают нас к базовому случаю.

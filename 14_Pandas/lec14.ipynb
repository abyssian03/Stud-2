#### Агрегации

Агрегация - это важный способ получать новые данные при анализе. Как мы уже знаем из numpy, в результате агрегации можно получить одно значение, а можно набор значений по осям.

Когда мы говорим, что агрегируем данные по осям, это означает, что мы их по сути группируем. В pandas есть разные способы группировки данных. Самый простой - это использовать метод .groupby().

Но groupby позволяет сгруппировать данные только по какому-то одному столбцу (признаку). Обычно к результату метода применяется .mean() или .sum().

```python
titanic.groupby('Sex').mean()
```

Эти методы можно применять и к отдельному столбцу, и к таблице целиком. Но если в таблице есть не только числовые данные, лучше считать по столбцам с помощью groupby, а для таблицы целиком используйте тогда describe().

```python
titanic.Age.mean()
```

```python
titanic.Survived.sum()
```

Также в pandas есть готовый метод OHE (One Hot Encoding), про который можно почитать на [вики](https://en.wikipedia.org/wiki/One-hot). Он полезен, если у нас есть колонка с категориальным признаком. Что такое категориальный признак? Даже если он из чисел, в отличие от цифрового, эти числа между собой арифметически не соотносятся. Чаще, впрочем, категориальный признак - это что-то вроде "из какого города": у нас есть n городов, и мы не можем их просто пронумеровать, потому что если мы укажем, что Москва у нас - 1, а Петербург - 2, то разве Петербург больше Москвы? Что это будет значить? В этом случае можно применить OHE в pandas к ней и получить дополнительные колонки:

```python
new_cols = pd.get_dummies(df.column_name, prefix='prefix', drop_first=True)
```

В Титанике нет удобных столбцов с признаками, которые хотелось бы заменить, поэтому код абстрактный. То есть, в строке выше мы создаем дополнительную табличку из n - 1 колонок для n вариантов значений колонки column_name с префиксом (названия колонок будут prefix_1, prefix_2...), и удаляем сразу первый столбец, потому что он избыточный для OHE.

Дальше можно смержить получившееся:

    df.drop('column_name', inplace=True) # исходный столбец больше не нужен
    df = df.join(new_cols)

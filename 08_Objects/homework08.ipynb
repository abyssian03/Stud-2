#### Задача 5.

Разберите код ниже. Подумайте, что означает каждый из атрибутов создаваемого там класса и почему была выбрана именно такая структура данных. Попробуйте разобраться в том, как работает магический метод .__str__(). Результаты своих размышлений напишите в комментариях.


```python
class Token:
    def __init__(self, text, lemma, pos_tag, morph_features):
        self.text = text
        self.lemma = lemma
        self.pos_tag = pos_tag
        self.morph_features = morph_features

    def __str__(self):
        return f"{self.text} ({self.lemma}) — {self.pos_tag}"

word1 = Token("коты", "кот", "NOUN", {"number": "plural", "case": "nominative"})
print(word1)
print(word1.morph_features["case"])
```
